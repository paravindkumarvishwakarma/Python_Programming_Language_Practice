### PY-CON-012 · Why 0.1 + 0.2 is not 0.3

`💡 Conceptual` · 🔴 Hard · **Topic:** Variables & Data Types

**Question**

Explain why `0.1 + 0.2 == 0.3` is `False` in Python. How would you handle money safely?

✍️ **Your answer:** *(double-click this cell and type your explanation)*



<details>
<summary>✅ Show model answer</summary>

Computers store decimals in **binary** (IEEE 754). Numbers such as `0.1` cannot be represented exactly in binary, just like `1/3` cannot be written exactly in decimal. The tiny errors add up: `0.1 + 0.2 = 0.30000000000000004`.

Safer approaches:

- Round before comparing: `round(0.1 + 0.2, 2) == 0.3`.
- Store money as **integers** (paise or cents) and divide only when displaying.
- Use the `decimal` module for exact decimal arithmetic.

</details>

### PY-CON-013 · Short-circuit evaluation

`💡 Conceptual` · 🔴 Hard · **Topic:** Operators

How do `and` / `or` work in Python? What do `0 or "default"` and `5 and 10` return, and why is it useful?

✍️ **Your answer:** *(double-click this cell and type your explanation)*



<details>
<summary>✅ Show model answer</summary>

`and` / `or` return **one of their operands**, not just `True` / `False`, and they stop (short-circuit) as soon as the answer is known.

### PY-CON-015 · Precedence and associativity

`💡 Conceptual` · 🟡 Medium · **Topic:** Operators

**Question**

What is operator precedence? In which order are `+ - * / // % **`, comparisons, `not`, `and` and `or` evaluated? Why is `2 ** 3 ** 2` equal to `512`?

✍️ **Your answer:** *(double-click this cell and type your explanation)*



<details>
<summary>✅ Show model answer</summary>

Precedence decides which operator runs first. From highest to lowest:

1. `**`
2. unary `+x`, `-x`, `~x`
3. `*`, `/`, `//`, `%`
4. `+`, `-`
5. shifts `<<`, `>>`, then `&`, `^`, `|`
6. comparisons, `is`, `in`
7. `not`
8. `and`
9. `or`

Most operators work left to right, but `**` is **right-associative**: `2 ** 3 ** 2 = 2 ** (3 ** 2) = 2 ** 9 = 512`.

When in doubt, add brackets: they make the intention obvious.

</details>

principal = float(input("Amount invested (₹): "))
